# 📄 Gerador de Orçamento em PDF — v3

Automatiza a criação de orçamentos de serviço cobrados por hora:

1. **Coleta** os dados do projeto (cliente, descrição, horas, valor/hora, prazo)
2. **Valida** as entradas, aceitando números no padrão brasileiro (`350,50` ou `1.234,56`)
3. **Calcula** o valor total
4. **Gera** um PDF com layout próprio, desenhado em código, com valores em **R$ 122.500,00**

Toda a lógica fica no módulo **`orcamento.py`**. Este notebook só conduz o passo a passo.

## 🆕 O que mudou em relação à versão 2.0

## 📚 2. Importação do módulo

```python
from orcamento import (
    ErroDeValidacao, criar_orcamento, gerar_pdf, resumo_texto
)

print("✅ Módulo carregado!")
```

## 📝 3. Entrada de dados

Responda às perguntas. Horas e valor da hora aceitam vírgula ou ponto (ex.: `350` ou `350,50`).

```python
def coletar_orcamento():
    """Pergunta os dados ao usuário e devolve um Orcamento validado (ou None)."""
    try:
        return criar_orcamento(
            cliente=input("👤 Nome do cliente (opcional): "),
            projeto=input("📌 Descrição do projeto: "),
            horas=input("⏱️  Horas previstas: "),
            valor_hora=input("💰 Valor da hora (R$): "),
            prazo=input("📅 Prazo estimado (ex.: 3 meses): "),
        )
    except ErroDeValidacao as erro:
        print(f"❌ {erro}")
        return None


orcamento = coletar_orcamento()
if orcamento:
    print("\n" + resumo_texto(orcamento))
```

## 📄 4. Gerar o PDF

```python
if orcamento:
    arquivo = gerar_pdf(orcamento, pasta_saida="orcamentos")
    print(f"✅ Orçamento salvo em: {arquivo}")
else:
    print("⚠️  Preencha os dados na etapa 3 antes de gerar o PDF.")
```
